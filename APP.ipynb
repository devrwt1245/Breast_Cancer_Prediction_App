```python
import streamlit as st
import pandas as pd
import numpy as np
import pickle


# ---------------------------------------------------------
# PAGE CONFIGURATION
# ---------------------------------------------------------

st.set_page_config(
    page_title="AI Powered House Price Prediction",
    page_icon="🏚️",
    layout="wide"
)


# ---------------------------------------------------------
# CUSTOM CSS
# ---------------------------------------------------------

st.markdown("""
<style>

.main {
    background: #f5f8ff;
}

.stButton > button {
    background: #0099ff;
    color: white;
    border-radius: 12px;
    height: 50px;
    width: 100%;
    font-size: 18px;
    font-weight: bold;
    border: none;
}

.stButton > button:hover {
    background: #007acc;
    color: white;
}

h1 {
    color: #003366;
}

h2 {
    color: #003366;
}

h3 {
    color: #003366;
}

.metric {
    font-size: 30px;
}

div[data-testid="stNumberInput"] input {
    border-radius: 8px;
}

</style>
""", unsafe_allow_html=True)


# ---------------------------------------------------------
# TITLE
# ---------------------------------------------------------

st.title("🏚️ Boston House Price Prediction Powered by AI")

st.write(
    "Here we are predicting house prices using Machine Learning."
)


# ---------------------------------------------------------
# LOAD MACHINE LEARNING MODEL
# ---------------------------------------------------------

try:

    with open("model.pkl", "rb") as file:
        model = pickle.load(file)

    with open("columns.pkl", "rb") as file:
        columns = pickle.load(file)

except FileNotFoundError:
    st.error(
        "❌ model.pkl or columns.pkl was not found. "
        "Please make sure both files are in the same folder as app.py."
    )
    st.stop()

except Exception as e:
    st.error(f"❌ Error loading the model: {e}")
    st.stop()


# ---------------------------------------------------------
# INPUT DATA
# ---------------------------------------------------------

data = {}

st.subheader("🏠 Enter House Details")

col1, col2, col3 = st.columns(3)

for i, col in enumerate(columns):

    if i % 3 == 0:

        with col1:
            data[col] = st.number_input(
                label=col,
                value=0.0,
                step=0.01
            )

    elif i % 3 == 1:

        with col2:
            data[col] = st.number_input(
                label=col,
                value=0.0,
                step=0.01
            )

    else:

        with col3:
            data[col] = st.number_input(
                label=col,
                value=0.0,
                step=0.01
            )


# ---------------------------------------------------------
# PREDICTION
# ---------------------------------------------------------

st.write("")

if st.button("🔮 Predict House Price"):

    try:

        # Convert input data into DataFrame
        df = pd.DataFrame([data])

        # Make prediction
        prediction = model.predict(df)[0]

        # Display prediction
        st.success(
            f"🏠 Estimated House Price: ${prediction:,.2f}"
        )

        st.balloons()

    except Exception as e:

        st.error(
            f"❌ Prediction Error: {e}"
        )


# ---------------------------------------------------------
# DIVIDER
# ---------------------------------------------------------

st.divider()


# ---------------------------------------------------------
# PROJECT HIGHLIGHTS
# ---------------------------------------------------------

st.subheader("🚀 Project Highlights")

st.markdown("""
### Machine Learning
- Linear Regression
- Ridge Regression
- Lasso Regression
- ElasticNet

### Data Preprocessing
- Data Cleaning
- Feature Engineering
- Outlier Detection
- StandardScaler
- MinMaxScaler

### Model Evaluation
- Adjusted R²
- Model Performance Evaluation

### Deployment
- Streamlit Web Application
- Interactive House Price Prediction
""")


# ---------------------------------------------------------
# FOOTER
# ---------------------------------------------------------

st.divider()

st.caption(
    "🤖 AI Powered House Price Prediction | "
    "Built with Python, Machine Learning & Streamlit"
)
```
